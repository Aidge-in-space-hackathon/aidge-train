Un graphique permet de **comprendre** les données.


```python

import matplotlib.pyplot as plt

plt.figure()
plt.scatter(df["sepal_length"], df["sepal_width"])
plt.xlabel("sepal_length")
plt.ylabel("sepal_width")
plt.title("Iris dataset")
plt.show()
```


### ✅ TODO 9.1 — Graphique par espèce


```python

# TODO 9.1
plt.figure()
for sp in df["species"].unique():
    d = df[df["species"] == sp]
    plt.scatter(d["sepal_length"], d["sepal_width"], label=sp)

plt.legend()
plt.show()
```


🔵 **Niveau 3 — Bonus (pré-ML)**

1. Normalisez les variables
2. Comparez les distances entre espèces
3. Lesquelles sont les plus proches ?


```python

# Niveau 3
from sklearn.preprocessing import StandardScaler
import numpy as np
from itertools import combinations

X = df.drop(columns=["species"])
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

means = pd.DataFrame(X_scaled, columns=X.columns).groupby(df["species"]).mean()

distances = {}
for a, b in combinations(means.index, 2):
    distances[(a, b)] = np.linalg.norm(means.loc[a] - means.loc[b])

distances
```
